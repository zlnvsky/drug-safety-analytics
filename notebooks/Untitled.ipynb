## 1.6 Create Features

```python
age_multipliers_days = {
    'YR': 365,
    'MON': 30,
    'WK': 7,
    'DY': 1,
    'DEC': 3650,
    'HR': 1/24
}
# age converted to days using age_cod multipliers
df['age_days'] = df['age'] * df['age_cod'].map(age_multipliers_days)
df['age_days'] = df['age_days'].round(0).astype('Int64')
```

```python
age_multipliers_years = {
    'YR': 1,
    'MON': 1/12,
    'WK': 1/52,
    'DY': 1/365,
    'DEC': 10,
    'HR': 1/8760
}
# age converted to years using age_cod multipliers
df['age_years'] = df['age'] * df['age_cod'].map(age_multipliers_years)
df['age_years'] = df['age_years'].round(0).astype('Int64')
```

```python
bins = [0, 28, 730, 4383, 6575, 23725, float('inf')]
labels = [
    'Neonate',
    'Infant',
    'Child',
    'Adolescent',
    'Adult',
    'Elderly'
]

df['age_grp'] = pd.cut(df['age_days'], bins=bins, labels=labels, right=False, ordered=False)
df['age_grp'] = df['age_grp'].cat.add_categories('UNK')
df['age_grp'] = df['age_grp'].fillna('UNK')

df['age_grp'].unique()
```

# 3. Indication dataframe cleaning

```python
def rename_columns(df):
    """
    Rename and standardize column names in a dataframe.
    """
    df = df.copy()
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(r'\s+', '_', regex=True)
        .str.replace(r'[^\w]', '', regex=True)
    )
    return df
```

```python
def normalize_str_values(df):
    """
    Normalize string values: uppercase, strip, remove special characters.
    """
    df = df.copy()
    str_cols = df.select_dtypes(['str', 'object']).columns
    df[str_cols] = df[str_cols].apply(lambda x: x
        .str.replace('.', '', regex=False)
        .str.replace('-', ' ', regex=False)
        .str.replace(r'\s+', ' ', regex=True)
        .str.strip()
        .str.upper())
    return df
```

```python
UNKNOWN_VALUES = ['UNKNOWN', 'UNKNWON', 'NOT SPECIFIED', 'NOT REPORTED',
                  'N/A', 'NA', 'NONE', 'NR', '-', '', ' ']

def standardize_unknown(df):
    """
    Replace common unknown/missing string values with UNK.
    """
    df = df.copy()
    str_cols = df.select_dtypes(['str', 'object']).columns
    df[str_cols] = df[str_cols].replace(UNKNOWN_VALUES, 'UNK')
    df[str_cols] = df[str_cols].apply(
        lambda x: x.str.replace(r'.*UNKNOWN.*', 'UNK', regex=True)
    )
    return df
```

```python
def convert_outc_cod(df):
    """
    Convert outc_cod to ordered categorical dtype.
    """
    df = df.copy()
    outc_order = ['OT', 'RI', 'CA', 'DS', 'HO', 'LT', 'DE']
    role_cats = pd.CategoricalDtype(outc_order, ordered=True)
    df['outc_cod'] = df['outc_cod'].astype(role_cats)
    return df
```
